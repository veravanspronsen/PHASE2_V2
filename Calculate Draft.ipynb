```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
mass = [43* 10 **3]  # kg

radius = 2.25/2
A = 3 * np.pi * radius**2 # m^2
rho = 1025 # kg/m^3

```

```python
for i in range(len(mass)):
    draft = mass[i] / (rho * A)
    volume_displaced = mass[i] / rho
    print(f"Draft: {draft:.2f} m and volume: {volume_displaced:.2f} m³ for mass {mass[i]/1000:.2f} tons")
```

```python
draft = [2.11, 3.60]
radius = [2.25/2, 1.82/2]
```

```python
for i in range(len(draft)):
    for j in range(len(radius)):
        onderkant = 3 * np.pi * radius[j]**2
        zijkant = draft[i] * 2 * radius[j] * np.pi
        totaal = onderkant + zijkant
        print(f"Total wetted surface area: {totaal:.2f} m² for draft {draft[i]:.2f} m and diameter {2 * radius[j]:.2f} m")
```
