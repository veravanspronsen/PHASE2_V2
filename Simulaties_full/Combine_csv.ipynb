```python
"""
Combine three sweep CSVs into one file.

Expects all three CSVs in the same folder with identical columns:
  Hs, Tp, Direction, damp_config, heave_std, heave_max, pitch_std, pitch_max, roll_std, roll_max
"""

import pandas as pd
from pathlib import Path
```

```python

# ── Config ──────────────────────────────────────────────────────────────────
DIR_RESULTS = Path('results')
DIR_RESULTS.mkdir(exist_ok=True)

FILES = [
    'C:\\Users\\verav\\Desktop\\Studie\\Afstuderen\\PHASE2_v2\\PHASE2_V2\\Simulaties_sweep\\results\\resultscombined_sweep.csv',
    'C:\\Users\\verav\\Desktop\\Studie\\Afstuderen\\PHASE2_v2\\PHASE2_V2\\Simulaties Theave\\results\\resultscombined_Theave.csv',
    'C:\\Users\\verav\\Desktop\\Studie\\Afstuderen\\PHASE2_v2\\PHASE2_V2\\Simulaties Tpitch\\results\\resultscombined_Tpitch.csv',
]

OUTPUT_FILE = 'resultscombined_all.csv'

# ── Load and concatenate ────────────────────────────────────────────────────
dfs = []
for fname in FILES:
    path = DIR_RESULTS / fname
    if not path.exists():
        print(f'⚠️  Not found: {path}')
        continue
    df = pd.read_csv(path)
    print(f'✓ Loaded {fname:<35} ({len(df):>4} rows, {df.shape[1]} cols)')
    dfs.append(df)

combined = pd.concat(dfs, axis=0, ignore_index=True)

# ── Save ────────────────────────────────────────────────────────────────────
out_path = DIR_RESULTS / OUTPUT_FILE
combined.to_csv(out_path, index=False)

print(f'\n✅ Combined {len(dfs)} files → {out_path}')
print(f'   Total rows: {len(combined)}')
print(f'   Columns   : {combined.columns.tolist()}')
```
